```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/superstore_raw.csv",
                 encoding="latin1")
```

```python
df.head()
```

```python

df.shape
df.info()
```

```python
df.columns =(df.columns
             .str.lower()
             .str.replace(" ","_")
             .str.replace("-","_")
             .str.strip()
            )
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.drop(columns=["postal_code"], inplace=True)
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["ship_date"] = pd.to_datetime(df["ship_date"])
```

```python
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
df["order_month_name"] = df["order_date"].dt.month_name()
```

```python
df[df["sales"] <= 0]
df[df["quantity"] <= 0]
```

```python
df.duplicated().sum()
```

```python
Q1 = df["sales"].quantile(0.25)
Q3 = df["sales"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
```

```python
df["is_outlier"] = ((df["sales"] < lower) | (df["sales"] > upper))
```

```python
df["sales"].sum()
df["profit"].sum()
```

```python
df["profit_margin"] = df["profit"]/df["sales"]
```

```python
df.head()
```

```python
df.to_csv("../data/processed/superstore_cleaned.csv", index=False)
```
